```python
import cv2 
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('face.jpg')
```

```python
img[0][0]
```

```python
cv2.imshow('angel', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
im2=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(im2)
plt.show()
```

```python
img_cr=im2[50:450,125:575]
```

```python
plt.imshow(img_cr)
plt.show()
```

```python
img_cr.shape
```

```python
size=(100,100)
img_resize=cv2.resize(img_cr, size, interpolation=cv2.INTER_AREA)
plt.imshow(img_resize)
plt.show()
```

```python
img_gray=cv2.cvtColor(img_cr,cv2.COLOR_RGB2GRAY)
plt.imshow(img_gray)
plt.show()
```

```python
cv2.imshow('angel', img_gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
_,img_tr = cv2.threshold(img_gray,150,255,0 )
cv2.imshow('angel', img_tr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_blr=cv2.GaussianBlur(img_cr, (51,51), 0)
cv2.imshow('angel', img_blr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
